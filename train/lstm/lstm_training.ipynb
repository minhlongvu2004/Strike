```python
import os 
import numpy as np
from sklearn.model_selection import train_test_split
from tensorflow.keras.utils import to_categorical
```

```python
dir = "./actions" # CHANGE YOUR PATH HERE
label_map = [label for label in os.listdir(dir)]
```

```python
data, labels, label_map = [], [],[]
for index, act in enumerate(os.listdir(dir)):
    label_map.append(act)
    for vid in os.listdir(f"{dir}/{act}"):
        vid_frames = []
        for frame in os.listdir(f"{dir}/{act}/{vid}"):
            frame_path = f"{dir}/{act}/{vid}/{frame}"
            frame_data = np.load(frame_path)
            vid_frames.append(frame_data)
        data.append(vid_frames)
        labels.append(index)
        
X = np.array(data)
Y = to_categorical(labels).astype(int)
```

```python
to_categorical(labels).astype(int)
```

```python
X_train, X_val, Y_train, Y_val = train_test_split(X, 
                                                    Y, 
                                                    test_size=0.05,
                                                    random_state=42,    # Makes the split reproducible
                                                    shuffle=True)
```

# Build LSTM

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping
```

```python

# Layer 1
model = Sequential()

# Input: (20 frames, 21 points * 6 =  features)
model.add(LSTM(128, 
               return_sequences=True, 
               activation='tanh',
               input_shape=(20, 126), dropout=0.2))


model.add(LSTM(64,
               activation='tanh', 
               return_sequences=False, 
               dropout=0.2))

model.add(Dense(64, activation='relu'))
model.add(Dropout(0.3))

model.add(Dense(32, activation='relu'))
model.add(Dropout(0.3))
model.add(Dense(len(label_map), activation='softmax'))
```

```python
model.compile(optimizer='Adam', 
              loss='categorical_crossentropy', 
              metrics=['categorical_accuracy'])
early_stopping = EarlyStopping(
    monitor='val_loss',
    patience=30,
    restore_best_weights=True,
    verbose=1
)
```

```python
model.fit(X_train, 
          Y_train, 
          validation_data = (X_val, Y_val),
          epochs=300
          )
```

```python
model.save('action.h5')

# load 
# model.load_weights('action.h5')
```
